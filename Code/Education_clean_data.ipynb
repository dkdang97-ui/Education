## Data Preparation
Select relevant subsets of the data

## The school information dataset includes many columns. We only need the year, school identity, location, school style information
Keep SCHOOL_YEAR, NCESSCH, LSTATE, LZIP, SCH_TYPE_TEXT, LEVEL, CHARTER_TEXT

```python
school_information=school_information[['SCHOOL_YEAR', 'NCESSCH', 'LSTATE', 'LZIP', 'SCH_TYPE_TEXT', 'LEVEL', 'CHARTER_TEXT']]
```

## Rename Columns

```python
edgap=edgap.rename(
    columns={
        "NCESSCH School ID":"id",
        "CT Unemployment Rate":"rate_unemployment",
        "CT Pct Adults with College Degree":"percent_college",
        "CT Pct Childre In Married Couple Family":"percent_married",
        "CT Median Household Income":"median_income",
        "School ACT average (or equivalent if SAT score)":"average_act",
        "School Pct Free and Reduced Lunch":"percent_lunch"
    }
)
```

```python
school_information=school_information.rename(
    columns={
        "SCHOOL_YEAR":"year",
        "NCESSCH":"id",
        "LSTATE":"state",
        "LZIP":"zip_code",
        "SCH_TYPE_TEXT":"school_type",
        "LEVEL":"school_level",
        "CHARTER_TEXT":"charter"
    }
)
```

## Join Dataframe

# 'ID' is the key identifier. Change data type for 'id' in both dataset from int to object(str)

```python
school_information['id']=school_information['id'].astype('object')
edgap['id']=edgap['id'].astype('object')
```

# Merge two datasets with edgap as the primary one.

```python
df=edgap.merge(
    school_information,
    how='left',
    on='id'
)
```

## Data Imputation

Define the predictor variables

```python
predictor_variables = [
    'rate_unemployment',
    'percent_college',
    'percent_married',
    'median_income',
    'percent_lunch',
    'state',
    'charter'
]
```

# Use iterative imputer to replace missing values

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
imputer = IterativeImputer()
```

```python
numerical_predictors = df[predictor_variables].select_dtypes(include='number').columns.to_list()
print(numerical_predictors)
```

Fit the imputer

```python
imputer.fit(df.loc[:,numerical_predictors])
```

```python
df.loc[:,numerical_predictors] = imputer.transform(df.loc[:,numerical_predictors])
```

Check for missing value 

```python
df.isna().sum().to_frame(name='Number of Missing Values')
```

Export the clean dataset

```python
df.to_csv(
    'education_clean.csv',
    encoding='utf-8-sig',
    index=False
)
```
